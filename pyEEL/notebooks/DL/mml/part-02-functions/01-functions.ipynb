## Functions in theory

````python
import numpy as np


def cross_entropy_loss(X, y):
    """
    Args:
        X: numpy.ndarray of shape (n_batch, n_dim).
            Contains the predictions in form of a probability distribution.
        y: numpy.ndarray of shape (n_batch, 1).
            Contains class labels for each data point in X.
        
    Returns:
        loss: numpy.float. 
            Cross entropy loss of the predictions.
    """
    
    exp_x = np.exp(X)
    probs = exp_x / np.sum(exp_x, axis=1, keepdims=True)
    log_probs = - np.log([probs[i, y[i]] for i in range(len(probs))])
    loss = np.mean(log_probs)
    
    return loss
````

### Functions as callable objects

````python
def linear(x, a, b):
    return a*x + b
````

````python
class Linear:
    def __init__(self, a, b):
        self.a = a
        self.b = b
        
    def __call__(self, x):
        return self.a*x + self.b
````

````python
f = Linear(2, -1) # this represents the function f(x) = 2*x - 1
f(2.1)
````

This code demonstrates how to create and use a custom Linear class to represent and evaluate linear functions. The first line creates an instance of the Linear class with parameters a=2 and b=-1, which mathematically represents the linear function f(x) = 2x - 1. The comment helpfully clarifies this relationship between the constructor parameters and the resulting mathematical function.

The second line shows the class being used in a very intuitive way by calling f(2.1). This works because the Linear class implements the __call__ method, which makes instances of the class "callable" like regular functions. When you call f(2.1), Python automatically invokes the __call__ method with x=2.1, which then computes and returns: 
```python
self.a * x + self.b = 2 * 2.1 + (-1) = 4.2 - 1 = 3.2.
```

This design pattern is particularly elegant because it allows objects to behave like functions while still maintaining the benefits of object-oriented programming, such as storing state (the parameters a and b) and providing additional methods like parameters() for introspection. This approach is commonly used in machine learning frameworks where you want mathematical functions to be both callable and able to store learnable parameters.

````python
class Linear:
    def __init__(self, a, b):
        self.a = a
        self.b = b
        
    def __call__(self, x):
        return self.a*x + self.b
    
    def parameters(self):
        return {"a": self.a, "b": self.b}
````

````python
f = Linear(2, -1)
f.parameters()
````

````python
class Linear:
    def __init__(self, a, b):
        self.a = a
        self.b = b
        
    def __call__(self, x):
        return self.a*x + self.b
    
    def __repr__(self):
        return f"Linear(a={self.a}, b={self.b})"
    
    def parameters(self):
        return {"a": self.a, "b": self.b}
````

### The `Function` base class

````python
class Function:
    def __init__(self):
        pass
    
    def __call__(self, *args, **kwargs):
        pass
    
    def parameters(self):
        return dict()
````

````python
import numpy as np


class Sigmoid(Function):            # the parent class is explicitly declared
    def __call__(self, x):
        return 1/(1 + np.exp(-x))
````

````python
sigmoid = Sigmoid()
````

````python
sigmoid(2)
````

````python
sigmoid.parameters()
````

### Composition in the object-oriented way

````python
composed = compose(Linear(2, -1), Sigmoid())
````

````python
composed(2)
````

````python
isinstance(composed, Function)
````

````python
composed.parameters()
````

````python
class Composition(Function):
    def __init__(self, *functions):
        self.functions = functions
        
    def __call__(self, x):
        
        for f in reversed(self.functions):
            x = f(x)
            
        return x
````

````python
composed = Composition(Linear(2, -1), Sigmoid())
````

````python
composed(2)
````

````python
composed.parameters()
````

````python
isinstance(composed, Function)
````
