## 1. What is USD?

### 🎬 Universal Scene Description (USD)

**USD** is an open-source framework developed by Pixar Animation Studios for describing, composing, simulating, and collaborating on 3D scenes. It was released as open source in 2016 and has become the industry standard for 3D data interchange.

### Key Features:

#### 🔹 **Composition**
USD's most powerful feature is its ability to **non-destructively** combine multiple layers of scene description. Think of it like layers in Photoshop, but for 3D scenes:
- **Layering**: Stack modifications without changing source data
- **Referencing**: Include external USD files
- **Variants**: Define alternatives (e.g., different costumes)
- **Payloads**: Lazy-load heavy data

#### 🔹 **Scalability**
Designed to handle massive scenes:
- Feature film-scale complexity (billions of primitives)
- Lazy loading and streaming
- Efficient memory usage

#### 🔹 **Extensibility**
- Custom schemas for domain-specific data
- Plugin architecture
- Language bindings (Python, C++)

#### 🔹 **Time-Varying Data**
- Native support for animation
- Frame-accurate sampling
- Velocity and acceleration data

### Why USD?

**Traditional 3D Formats:**
```
OBJ     → Simple, but no animation, materials limited
FBX     → Proprietary, complex, inconsistent implementations
glTF    → Modern, but focused on real-time/web delivery
Alembic → Great for caches, but not a full pipeline format
```

**USD Advantages:**
```
✅ Open source and well-documented
✅ Handles entire production pipelines
✅ Non-destructive workflow
✅ Collaborative (multiple artists, same scene)
✅ Industry adoption (Pixar, NVIDIA, Apple, Adobe, Autodesk)
✅ Supports both real-time and offline rendering
```

### USD in Production:

- **Film**: Pixar, Disney, ILM, Weta Digital
- **Games**: NVIDIA Omniverse, Unity, Unreal Engine
- **AR/VR**: Apple (USDZ for AR QuickLook)
- **Architecture**: Autodesk, Trimble
- **Simulation**: NVIDIA Isaac Sim

### USD Ecosystem:

```
OpenUSD Core
    ├── UsdGeom    (Geometry primitives)
    ├── UsdShade   (Material networks)
    ├── UsdLux     (Lights)
    ├── UsdSkel    (Skeletal animation)
    ├── UsdPhysics (Physics simulation)
    ├── UsdVol     (Volumes)
    └── Custom Schemas (Extensible)
```

### Helper Functions for Visualization

````python
def print_stage_tree(stage, prim=None, indent=0):
    """
    Print the hierarchy of prims in a USD stage.
    
    Args:
        stage: USD stage to print
        prim: Current prim (None = start from root)
        indent: Current indentation level
    """
    if prim is None:
        prim = stage.GetPseudoRoot()
    
    for child in prim.GetChildren():
        type_name = child.GetTypeName()
        print("  " * indent + f"└─ {child.GetName()} ({type_name})")
        print_stage_tree(stage, child, indent + 1)

def save_and_print(stage, filepath):
    """
    Save a USD stage and print both the file path and USDA content.
    
    Args:
        stage: USD stage to save
        filepath: Output file path
    """
    # Save the stage
    stage.GetRootLayer().Export(str(filepath))
    print(f"\n💾 Saved: {filepath}")
    
    # Print the hierarchy
    print("\n📊 Scene Hierarchy:")
    print_stage_tree(stage)
    
    # Print the USDA content if file size is reasonable
    if os.path.getsize(filepath) < 10000:  # Less than 10KB
        print("\n📄 USDA Content:")
        print("-" * 60)
        with open(filepath, 'r') as f:
            print(f.read())
        print("-" * 60)
    else:
        print("\n(File too large to display - use usdview or text editor)")

def open_in_usdview(filepath):
    """
    Open a USD file in usdview (if available).
    
    Args:
        filepath: Path to USD file
    """
    import subprocess
    try:
        print(f"\n🎬 Opening in usdview: {filepath}")
        subprocess.Popen(['usdview', str(filepath)])
    except FileNotFoundError:
        print("⚠️  usdview not found. Install with: pip install usd-core[usdview]")
        print(f"   You can manually open: {filepath}")

print("✅ Helper functions loaded!")
````
